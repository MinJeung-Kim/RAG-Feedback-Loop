# Streamlit 앱 실행

아래 셀을 실행하면 백그라운드에서 `app.py`가 Streamlit으로 뜹니다.
접속 URL을 브라우저에서 열어주세요.

```python
import os
import subprocess

# app.py / .venv 가 있는 프로젝트 폴더 (이 노트북이 있는 위치)
BASE = os.getcwd()

proc = subprocess.Popen(
    [
        os.path.join(BASE, ".venv", "Scripts", "streamlit.exe"), "run",
        os.path.join(BASE, "app.py"),
        "--server.port", "8501",
        "--server.enableCORS", "false",
        "--server.enableXsrfProtection", "false",
    ],
    cwd=BASE,
)

print("접속 URL: http://localhost:8501")
print(f"프로세스 PID: {proc.pid}")
```

## 앱 종료

위에서 띄운 Streamlit 프로세스를 멈추려면 아래 셀을 실행하세요.

```python
proc.terminate()
print("앱을 종료했습니다.")
```
